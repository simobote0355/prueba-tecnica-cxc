```python
import pandas as pd
import sys
import os

sys.path.append(os.path.abspath('../..'))
from src.modelo.data_loader import DataLoader
from src.modelo.feature_builder import FeatureBuilder
from src.modelo.model import Model
```

```python
# Cargar sábana
df = DataLoader("../../src/data/sabana_cxc.csv").cargar()
y = df['pago_total']

# Construir features
fb = FeatureBuilder(top_n_transacciones=10)
X = fb.fit_transform(df)

# Entrenar y evaluar
modelo = Model().entrenar(X, y)
metricas = modelo.evaluar()
print(metricas)

# Guardar modelo y métricas
modelo.guardar("../../src/modelo/modelo_ejemplo.pkl")
pd.DataFrame([metricas]).to_csv("../../src/metricas/metricas_modelo.csv", index=False)

# Generar probabilidad para todas las cuentas (no solo el test set)
df['prob_pago'] = modelo.predecir_proba(X)
df['valor_esperado_recuperar'] = df['vlr_pendiente_pago'] * df['prob_pago']
df['valor_en_riesgo'] = df['vlr_pendiente_pago'] * (1 - df['prob_pago'])
df.to_csv("../../src/data/sabana_cxc_con_prediccion.csv", index=False)
```
