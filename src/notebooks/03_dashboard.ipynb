```python
import pandas as pd
```

```python
df = pd.read_csv("../../src/data/sabana_cxc_con_prediccion.csv")

total_registros = len(df)
vlr_original_total = df['vlr_original'].sum()
vlr_pagado_total = df['vlr_pagado'].sum()
vlr_pendiente_total = df['vlr_pendiente_pago'].sum()
    
valor_riesgo_total = df['valor_en_riesgo'].sum()
valor_esperado_total = df['valor_esperado_recuperar'].sum()
    
prob_pago_promedio = df['prob_pago'].mean()
dias_mora_promedio = df['dias_hasta_ultimo_pago'].mean()
porc_recuperado_promedio = df['porc_recuperado'].mean()

print("Métricas  globales")
print(f"Total Cuentas Analizadas: {total_registros:,}")    
print(f"Monto Total Generado: ${vlr_original_total:,.2f}")
print(f"Monto Total Pagado: ${vlr_pagado_total:,.2f} ({vlr_pagado_total/vlr_original_total:.1%})")
print(f"Monto Pendiente de Pago: ${vlr_pendiente_total:,.2f} ({vlr_pendiente_total/vlr_original_total:.1%})")
print(f"Valor Esperado a Recuperar: ${valor_esperado_total:,.2f} ({valor_esperado_total/vlr_pendiente_total:.1%} del pendiente)")
print(f"Valor en Riesgo (Pérdida Exp.): ${valor_riesgo_total:,.2f} ({valor_riesgo_total/vlr_pendiente_total:.1%} del pendiente)")
print(f"Probabilidad Promedio de Pago: {prob_pago_promedio:.2%}")
print(f"Porcentaje Recuperado Promedio: {porc_recuperado_promedio:.2%}")
print(f"Promedio Días sin Pago: {dias_mora_promedio:.1f} días")
```

```python
alto_riesgo = df[df['prob_pago'] < 0.40]

num_cuentas = len(alto_riesgo)
monto_pendiente = alto_riesgo['vlr_pendiente_pago'].sum()
valor_en_riesgo = alto_riesgo['valor_en_riesgo'].sum()
valor_esperado = alto_riesgo['valor_esperado_recuperar'].sum()
prob_media = alto_riesgo['prob_pago'].mean()

porc_cuentas = (num_cuentas / len(df)) * 100
porc_pendiente = (monto_pendiente / df['vlr_pendiente_pago'].sum()) * 100
porc_riesgo = (valor_en_riesgo / df['valor_en_riesgo'].sum()) * 100

print(f"Cuentas: {num_cuentas:,} ({porc_cuentas:.2f}%)")
print(f"Saldo Pendiente: ${monto_pendiente:,.2f} ({porc_pendiente:.2f}%)")
print(f"Pérdida Esperada: ${valor_en_riesgo:,.2f} ({porc_riesgo:.2f}%)")
print(f"Probabilidad Promedio: {prob_media:.2%}")
```
